```python
import sympy as sym
```

```python
EI, q, L1, L2, M = sym.symbols('EI q L1 L2 M')

L1 = 4
L2 = 6

q = 25

phi_1 = M * L1 / EI / 3 + q * L1**3 / EI / 24
phi_2 = - M * L2 / EI / 3

display(phi_1, phi_2)
eq = sym.Eq(phi_1, phi_2)

sol = sym.solve(eq, M)[0]
display(sol)
```

```python
EI, q, L1, L2, A_v = sym.symbols('EI q L1 L2 A_v')

L1 = 4
L2 = 6
q = sym.nsimplify(25)
EI = sym.nsimplify(21.33333333333333333*3/6*5*1000)
#q = sym.nsimplify(25)
#EI = 80000*2/3

M_B = - q * L1**2 / 2 + A_v * L1
display(M_B)

phi_B = M_B * L2 / EI / 3
display(phi_B)
display(phi_B.evalf())
4
w_A = phi_B * L1 - q * L1**4 / 8 / EI + A_v * L1**3 / 3 / EI
display(w_A)
display(w_A.evalf())

eq = sym.Eq(w_A, 0)
sol = sym.solve(eq, A_v)[0]
display(sol)

display(M_B.subs(A_v, sol))
```
