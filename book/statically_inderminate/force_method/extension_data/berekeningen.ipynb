```python
import sympy as sym

L_1, L_2, L_3, q, N, Bv = sym.symbols('L_1 L_2 L_3 q N Bv')

EA = sym.symbols('EA')

L_1 = sym.nsimplify(4)
L_2 = sym.nsimplify(4)
L_3 = sym.nsimplify(5)
q = sym.nsimplify(28)
display(q)
eq1 = sym.Eq(Bv * (L_1 + L_2) - q * L_2 * (L_1 + L_2/2 ) + N * 4 / 5 * L_1 , 0)
N = sym.solve(eq1, N)[0]
display(N)
L_CD = L_1 / 3 * 5
delta_L = N * L_CD / EA
display(delta_L)
w_D = delta_L / 4 * 5
display(w_D.simplify().expand())
w_E = w_D / L_1 * (L_1 + L_2)
display(w_E.simplify().expand())
delta_L_BE = Bv * L_3 / EA
display(delta_L_BE)
w_B = w_E - delta_L_BE
display(w_B.simplify().expand())
eq2 = sym.Eq(w_B, 0)
B_v_sol = sym.solve(eq2, Bv)[0]
display(B_v_sol)
display(N.subs(Bv, B_v_sol))
display(w_D.subs(Bv, B_v_sol))
display(w_E.subs(Bv, B_v_sol))
```
